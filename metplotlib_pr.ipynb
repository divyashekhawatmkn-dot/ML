```python
import pandas as pd 

import matplotlib.pyplot as plt


df = pd.read_csv(r"C:\Users\Hp\OneDrive\Desktop\Advance python\cleaned_student_performance.csv")
df
```

```python
df.info()
```

```python
bar = plt.barh(df["Student Name"],df["Attendance_%"])
plt.bar_label(bar)
plt.xlabel("Attendance")
plt.ylabel("Students Name")
plt.title("Attendance of students")
plt.show()
```

```python
df["Gender"] = df["Gender"].fillna("Unknown")
```

```python
plt.bar(df["Gender"],df["Attendance_%"],color="pink",edgecolor="black")

plt.xlabel("Gender")
plt.ylabel("Attendance")
plt.title("Gender wise Attendance")
plt.show()
```

```python
plt.barh(df["Student Name"],df["Exam_Score"])

plt.xlabel("Marks")
plt.ylabel("Students Name")
plt.title("Marks of students")
plt.show()
```

```python
avg_marks = df.groupby("Class")["Exam_Score"].mean()

plt.plot(avg_marks.index,avg_marks.values,color="maroon",marker="^")
plt.xlabel("Class")
plt.ylabel("Marks of students")
plt.title("Class wise Report")
plt.show()
```

```python
avg_marks = df.groupby("Class")["Exam_Score"].mean()
avg_att = df.groupby("Class")["Attendance_%"].mean()

plt.plot(avg_marks.index,avg_marks.values,marker="o",color="pink",label="Marks")
plt.plot(avg_att.index,avg_att.values,marker="^",label ="Attendance")
plt.xlabel("Class")
plt.ylabel("Students")
plt.title("Class wise Report")
plt.legend()
plt.show()
```

```python
plt.scatter(df["Study_Hours_Per_Day"],df["Exam_Score"])
plt.xlabel("Study hour")
plt.ylabel("Marks")
plt.title("Relationship between study hour and marks")
plt.show()
```

```python
# Create a bar chart to display the number of students in each gender category.

gender_count = df["Gender"].value_counts()

fig = plt.figure(figsize=(6,5),facecolor="lightblue")
plt.bar(gender_count.index,gender_count.values,color="pink",edgecolor="black")
plt.title("Gneder wise no of students")
plt.xlabel("Gender")
plt.ylabel("Student")
plt.show()
```

```python
# 11. Create a histogram of students' marks using 10 bins. Interpret the distribution.



plt.hist(df["Exam_Score"],bins=10,edgecolor="black",color="purple")
plt.title("Interpret the distribution of marks")
plt.grid()
plt.show()
```

```python
# 12. Calculate the average marks for each subject and display the results in a bar chart.
s = df.groupby("Subject")["Exam_Score"].mean()

plt.bar(s.index,s.values,color=["pink","lightblue","purple",])
plt.title("Average marks of Subject")
plt.ylabel("Marks")
plt.xlabel("Subject")
plt.show()
```

```python
# 13. Create a horizontal bar chart using plt.barh() to display the number of students in each city.
sc = df.groupby("City")["Student_ID"].count()
sc

fig = plt.figure(figsize=(8,5),facecolor="pink")
plt.barh(sc.index,sc.values,color="purple",edgecolor="black")
plt.title("Students according to the citys")
plt.ylabel("Students")
plt.xlabel("Cities")
plt.show()
```

```python
# 14. Create a box plot of students' marks to identify potential outliers.

plt.boxplot(df["Exam_Score"],
            patch_artist=True,
            boxprops=dict(facecolor="pink"),
            medianprops=dict(color="red", linewidth=2),
            whiskerprops=dict(color="black"),
            capprops=dict(color="black"),
            flierprops=dict(marker="o", markerfacecolor="orange"))
plt.title("Students marks")
plt.grid()
plt.show()
```

```python
# 15. Create a line chart of students' marks and display the value of each data point on the chart.


plt.plot(df.index, df["Exam_Score"], marker="o")

for x, y in zip(df.index, df["Exam_Score"]):
    plt.text(x, y, str(y))

plt.title("Students' Marks")
plt.xlabel("Student Index")
plt.ylabel("Marks")
plt.show()
```

```python
# Section C: Subplots

# 16. Create two charts in a single figure: a bar chart of gender counts and a histogram of marks.


a = df["Gender"].value_counts()

fig = plt.figure(figsize=(10,5),facecolor="lightblue")
plt.subplot(1,2,1)
plt.bar(a.index,a.values,color=["pink","lightblue","purple"])
plt.title("Gender count")


plt.subplot(1,2,2)
plt.hist(df["Exam_Score"],color="pink")
plt.title("Marks distribution")

plt.show()
```

```python
# 17. Create two charts in a single figure:
# - First chart: Bar chart of average marks by subject.
# - Second chart: Histogram of marks.


avg = df.groupby("Subject")["Exam_Score"].mean()
fig = plt.figure(figsize=(10,6),facecolor="pink")
plt.subplot(1,2,1)
plt.bar(avg.index,avg.values,color=["pink","lightblue","purple"],edgecolor="black")
plt.title(" Average marks by subject")
plt.ylabel("Marks")
plt.xlabel("Subject")


plt.subplot(1,2,2)
plt.hist(df["Exam_Score"],bins=5,color="lightblue",edgecolor="black")
plt.title("Marks distribution")

plt.show()
```

```python
# 18. Use plt.subplots(1, 2, figsize=(12, 4)) to create:
# - First chart: Line plot of students' marks.
# - Second chart: Bar chart of student counts by city.


fig,ax = plt.subplots(1,2,figsize=(10,4),facecolor="pink")

ax[0].plot(df.index,df["Exam_Score"],marker="D",color="lightblue",markerfacecolor="purple")
ax[0].set_title("Students' marks.")

a = df["City"].value_counts()

ax[1].bar(a.index,a.values,color=["pink","lightblue","purple"])
ax[1].set_title("Student counts by city")
ax[1].set_xlabel("City")
ax[1].set_ylabel("Students")
plt.tight_layout()
plt.show()

```

```python
#  19. Use plt.subplots(2, 2) to create four charts:
# - Gender-wise student count.
# - Class-wise student count.
# - Histogram of marks.
# - Box plot of marks.


a = df["Gender"].value_counts()
b = df["Class"].value_counts()
fig ,ax = plt.subplots(2,2,figsize=(12,8),facecolor="lightblue")

#  - Gender-wise student count.
ax[0,0].bar(a.index,a.values,color=["pink","lightgreen","purple"],edgecolor="black")
ax[0,0].set_title("Gender-wise student count")

# - Class-wise student count.
ax[0,1].bar(b.index,b.values,color=["pink","#FF8B94","purple"],edgecolor="black")
ax[0,1].set_title("Class-wise student count")


# - Histogram of marks.
ax[1,0].hist(df["Exam_Score"],bins=5,color="lightgreen",edgecolor="black")
ax[1,0].set_title("Histogram of marks")


# - Box plot of marks.

ax[1,1].boxplot(df["Exam_Score"])
ax[1,1].set_title("Box plot of marks")


plt.grid()
plt.show()
```
