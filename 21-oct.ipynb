```python
import pandas as pd
import numpy as np
from statistics import mode

df = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Age": [18, 20, 19, 21, 22, 20, 19, 23, 21, 20],
    "Marks": [45, 60, 75, 80, 90, 60, 55, 95, 70, 70],
    "Study_Hours": [2, 4, 5, 6, 8, 3, 2, 9, 5, 6],
    "Attendance": [70, 80, 85, 90, 95, 75, 65, 98, 88, 82]
})

df
```

```python
# Q1. Calculate the mean of the Marks column.
print("Mean =",df["Marks"].mean())
```

```python
# Q2. Find the median of the Age column.
print("Median=",df["Marks"].median())
```

```python
# Q3. Find the mode of the Marks column.
print(df["Marks"].mode())
```

```python
# Q4. Find the maximum and minimum values of Study_Hours.
print("Maximum Hour=",df["Study_Hours"].max())
print("Minimum Hour=",df["Study_Hours"].min())
```

```python
# Q5. Calculate the range of Marks.
max = df["Marks"].max()
max

min = df["Marks"].min()
min

range = max - min
print("Range =",range)
```

```python
# Q6. Calculate the standard deviation of Marks using Pandas.
print("STD=",df["Marks"].std())
```

```python
# Q7. Calculate the variance of Marks using Pandas.
print("Variance =",df["Marks"].var())
```

```python
# Q8. Find the total marks of all students.
print("Total marks=",df["Marks"].sum())
```

```python
# Q9. Count how many students have marks greater than 70.
df[df["Marks"]>70]
```

```python
# Q10. Find the average attendance of all students.
print("Average Attendance =",df["Attendance"].mean())
```

```python
# Q11. Calculate the 25th percentile of Marks.
print("25th percentile =",df["Marks"].quantile(0.25))
```

```python
# Q12. Calculate the 75th percentile of Marks.
print("75th percentile =",df["Marks"].quantile(0.75))
```

```python
# Q13. Find the first quartile (Q1) and third quartile (Q3) of Marks.

Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)

print(Q1)
print(Q3)
```

```python
# Q14. Calculate the Interquartile Range (IQR).
IQR =  Q3-Q1
print(IQR)
```
