```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("titanic")
df
```

```python
# 1. Were there more male passengers or female passengers on the Titanic?

sns.countplot(
    x="sex",
    data=sns.load_dataset("titanic"),
    hue="sex",
    palette="Set2",
    order=["female","male"])
plt.show()




```

```python
# 2. Which passenger class had the highest number of passengers?
sns.countplot(
    x="pclass",
    data=sns.load_dataset("titanic"),
    palette="Set2")
plt.title("Number of Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
# 3. Which gender had a higher survival rate: males or females?

sns.barplot(
    x="sex",
    y="survived",
    data=sns.load_dataset("titanic"),
    hue="sex",
    palette="Set2",
    errorbar=None)
plt.show()
```

```python
# 4. Did first-class passengers have a higher chance of survival?
sns.barplot(
    x="class",
    y="survived",
    data=sns.load_dataset("titanic"),
    hue="class",
    errorbar=None)

plt.show()
```

```python
# 5. Did children have a higher survival rate than adults?
df["Age_Group"] = df["age"].apply(
    lambda x: "Child" if x < 18 else "Adult"
)

sns.barplot(
    x="Age_Group",
    y="survived",
    data=df,
    palette="Set2")

plt.show()
```

```python
# 6. Were passengers with more expensive tickets more likely to survive?
sns.barplot(x="survived",y="fare",data=df)
plt.show()
```

```python
# 7. Did passengers travelling alone have a lower chance of survival?
sns.barplot(x="alone",y="survived",data=df,palette="Set3",errorbar=None)
plt.title("Survival rate with alone")
plt.show()
```

```python
# 8. Which embarkation location had the highest passenger survival rate?
sns.barplot(
    x="embark_town",
    y="survived",
    data=df,
    palette="Set2"
)
plt.title("Survival Rate by Embarkation Location")
plt.xlabel("Embarkation Location")
plt.ylabel("Survival Rate")
plt.show()
```

```python
# 9. Did family size affect the chances of survival?
df["Family_Size"] = df["sibsp"] + df["parch"] + 1


sns.barplot(x="Family_Size",y="survived",data=df,palette="Set2")
plt.show()
```
