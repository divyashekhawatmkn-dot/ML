```python
# 3. Standardize Gender values into consistent categories.
df["Gender"].unique()

df["Gender"] = df["Gender"].replace({
    "MALE":"Male",
    "m":"Male",
    "female":"Female",
    "f":"Female"
})
print(df["Gender"])
```

```python
# 4. Standardize Subject names, such as Mathematics and math.
print(df["Subject"].unique())
```

```python
df["Subject"] = df["Subject"].replace({
    "english":"English",
    "math":"Maths",
    "Mathematics":"Maths",
    "SCIENCE":"Science"
})
print(df["Subject"])
```

```python
print(df["Subject"].unique())
```

```python
# 5. Standardize City names with consistent capitalization.
print(df["City"].unique())
```
