```python
# 1. Standardize missing-value markers such as N/A, unknown, blank strings and - wherever appropriate.


df.replace(["N/A", "Unknown","unknown", " ", "-"], 'nan', inplace=True)
df
```

```python
print(df.isnull().sum())
```

```python
df["Class"].unique()
df["Class"] = df["Class"].replace({
    "Ten": "10th",
    "Class 10":"10th",
    "XI":"11th",
    "Class 12":"12th",
    "11":"11th",
    "12":"12th",
    "10":"10th"

})
df["Class"].unique()
```

```python
# 3. Standardize Gender values into consistent categories.


df["Gender"] = df["Gender"].str.strip().str.lower()

df["Gender"] = df["Gender"].replace({
    "male": "Male",
    "m": "Male",
    "female": "Female",
    "f": "Female"
})
```

```python
print(df["Gender"].unique())
```

```python
# 4. Standardize Subject names, such as Mathematics and math.
print(df["Subject"].unique())
```

```python
df["Subject"] = df["Subject"].replace({
    "english":"English",
    "math":"Maths",
    "Mathematics":"Maths",
    "SCIENCE":"Science"
})
print(df["Subject"].unique())
```

```python
print(df["Subject"].unique())
```

```python
# 5. Standardize City names with consistent capitalization.
print(df["City"].unique())
```

```python
df["City"] = df["City"].replace({
    "jaipur":"Jaipur",
    "Jodhpur ":"Jodhpur",
    "SIKAR":"Sikar",
    "sikar":"Sikar"
})
```

```python
print(df["City"].unique())
```

```python
# 6. Convert Attendance_%, Study_Hours_Per_Day and Exam_Score into numeric data types.
col = ["Attendance_%", "Study_Hours_Per_Day","Exam_Score"]
for cols in col:
    df[cols] = pd.to_numeric(df[cols],errors="coerce")
```

```python
# 7. Identify attendance values outside 0–100, negative study hours and exam scores outside 0–100.

df[df["Attendance_%"].lt(0)|df["Attendance_%"].gt(100)]
```

```python
df[df["Study_Hours_Per_Day"]<0]
```

```python
df[df["Exam_Score"].lt(0)|df["Exam_Score"].gt(100)]
```

```python
# 8. Handle invalid numerical values using a clearly stated rule.
# Handle invalid attendance
df.loc[
    (df["Attendance_%"] < 0) | (df["Attendance_%"] > 100),
    "Attendance_%"
] = np.nan

# Handle negative study hours
df.loc[
    df["Study_Hours_Per_Day"] < 0,
    "Study_Hours_Per_Day"
] = np.nan

# Handle invalid exam scores
df.loc[
    (df["Exam_Score"] < 0) | (df["Exam_Score"] > 100),
    "Exam_Score"
] = np.nan




print(df["Attendance_%"])
```

```python
# 9. Remove exact duplicate rows, but do not automatically remove every repeated Student_ID.
df.drop_duplicates(inplace=True)

print("Remaining duplicate rows:", df.duplicated().sum())
```

```python
print(df["Exam_Date"].head(5))
```

```python
# 10. Convert Exam_Date into a proper datetime data type and identify invalid dates.
df["Exam_Date"] = pd.to_datetime(df["Exam_Date"],format="%d-%m-%Y",errors="coerce")
```

```python
print(df["Exam_Date"].head(5))
```

```python
# 11. Handle missing numerical and categorical values using suitable strategies.

num_cols = [
    "Attendance_%",
    "Study_Hours_Per_Day",
    "Exam_Score"
]

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

print(df.isnull().sum())
```

```python
cat_cols = [
    "Gender",
    "Class",
    "Subject",
    "City",
    "Parent_Contacted"
]

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])





df["Remarks"] = df["Remarks"].fillna("Not Provided")


print(df["Exam_Date"].isna().sum())

df["Exam_Date"] = df["Exam_Date"].fillna(pd.Timestamp("2025-01-15"))

print(df.isnull().sum())
```

```python
# 9. Create a new column named Performance_Level using a rule you define and document.
condition = [
    df["Exam_Score"] >= 90,
    df["Exam_Score"] >= 70,
    df["Exam_Score"] >= 60,
    df["Exam_Score"] >= 50

]

choice = [
    "A+",
    "A",
    "B",
    "C"
]

df["Performance_Level"] = np.select(condition,choice,default="Fail")
df
```
