```python
import matplotlib.pyplot as plt
```

```python
month = ["A","B","C","D","E","F"]
sales = [300,400,200,700,400,900]


plt.plot(month,sales,marker="o",color="pink")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthlt Sales")
plt.show()
```

```python
month = ["A","B","C","D","E","F"]
sales = [300,400,200,700,400,900]
marketing = [406,362,480,90,236,780]


plt.plot(month,sales,marker="o",color="pink",label="Sales")
plt.plot(month,marketing,marker="^",label="Marketing")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthlt Sales")

plt.legend()
plt.grid()
plt.show()
```

```python
category = ["Electronics","clothing","Furniture","Beauty","Sports"]
sales = [5000,8000,10000,6000,2000]
fig = plt.figure(figsize=(6, 4), facecolor='lightblue')
bar = plt.bar(category,sales,color="Pink",width=0.5,edgecolor="black")

plt.bar_label(bar)
plt.xlabel("Category")
plt.ylabel("Sales")
plt.title("Category wise sales")
plt.show()
```

```python
category = ["Electronics","clothing","Furniture","Beauty","Sports"]
sales = [5000,8000,10000,6000,2000]

plt.barh(category, sales)


plt.xlabel("Category")
plt.ylabel("Sales")
plt.title("Category wise sales")
plt.show()
```

```python
marks = [38,59,29,60,90,20,77,98,99,45,89,56,10,59]

plt.hist(marks,bins=6,edgecolor="black",color="pink")
plt.xlabel("Marks")
plt.ylabel("Distribution")
plt.title("Marks distribution")
plt.show()
```

```python
experince = [5,7,9,1,3,10,7,4,6]
salry = [50000,70000,90000,10000,30000,100000,70000,40000,60000]

plt.scatter(experince,salry,color="orange")
plt.xlabel("experince")
plt.ylabel("Salary")
plt.title("Relationship")
plt.grid(alpha=0.5)
plt.show()
```

```python
category = ["Electronics","clothing","Furniture","Beauty","Sports"]
sales = [5000,8000,10000,6000,2000]
explode = [0.1, 0.2, 0.5, 0, 0]

plt.pie(sales,labels=category, explode = explode, autopct="%1.1f%%",shadow=True)

plt.title("Category wise sales",fontsize=18)
plt.show()
```

```python
month = ["A","B","C","D","E","F"]
sales = [300,400,200,700,400,900]
profit = [406,362,480,90,236,780]


#subplot(row,colm,postion)

plt.subplot(1,2,1)
plt.plot(month,sales)
plt.title("Sales")

plt.subplot(1,2,2)
plt.bar(month,profit)
plt.title("Profit")

plt.show()

```

```python
import numpy as np

students = ["A", "B", "C"]

boys = [70, 80, 60]
girls = [75, 85, 65]

x = np.arange(len(students))
width = 0.35

plt.bar(x-width/2, boys, width, label="Boys")
plt.bar(x+width/2, girls, width, label="Girls")

plt.xticks(x, students)
plt.legend()
plt.show()
```

```python
students = ["A", "B", "C"]

boys = [40, 50, 60]
girls = [30, 35, 25]

plt.bar(students, boys, label="Boys")
plt.bar(students, girls, bottom=boys, label="Girls")

plt.legend()
plt.show()
```
