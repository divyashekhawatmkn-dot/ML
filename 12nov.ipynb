```python
import seaborn as sns

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    "Student": ["Aman", "Riya", "Karan", "Neha", "Rohit",
                "Pooja", "Arjun", "Simran", "Vikas", "Anjali",
                "Mohit", "Priya", "Rahul", "Sneha", "Varun"],

    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male"],

    "Study_Hours": [2, 5, 3, 6, 4, 7, 3, 8, 2, 6,
                    5, 7, 4, 9, 3],

    "Attendance": [65, 90, 72, 95, 80, 92, 70, 98, 60, 88,
                   85, 94, 76, 99, 68],

    "Marks": [45, 78, 55, 88, 65, 92, 58, 95, 40, 82,
              75, 90, 62, 98, 50],

    "Age": [18, 19, 18, 20, 19, 20, 18, 21, 19, 20,
            18, 21, 19, 20, 18],

    "Result": ["Fail", "Pass", "Fail", "Pass", "Pass",
               "Pass", "Fail", "Pass", "Fail", "Pass",
               "Pass", "Pass", "Pass", "Pass", "Fail"]
}

df = pd.DataFrame(data)
df
```

```python
sns.lineplot(x=df.index,y="Attendance",data=df,marker="o")

plt.xlabel("Students")
plt.show()
```

```python
a = df["Gender"].value_counts()

sns.barplot(x=a.index,y=a.values)
plt.show()
```

```python
sns.get_dataset_names()
```

```python
sns.barplot(
    x="Gender",
    y="Marks",
    data=df,
    hue="Gender",
    palette="Set3",  # ye bar ka color mai change krta h
    errorbar=None
)

plt.show()
```

```python
sns.boxplot(
    x="Gender",
    y="Marks",
    data=df,
    palette="Set2",
    showfliers=True,
)

plt.show()
```
